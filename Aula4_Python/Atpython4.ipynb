```python
import pandas as pd
import os

path = "CD2022_Populacao_2010_Compatibilizada_20231222.xlsx"
```

```python
# 4 - Leitura da tabela
bruto = pd.read_excel(path, header=None)
bruto.head(10)
```

```python
dados = bruto.iloc[3:5573].copy().reset_index(drop=True)
dados = dados.drop(columns=[0])

colunas = ["uf", "cod_uf", "cod_munic", "municipio", "pop_2010_sinopse", "pop_2010_compatibilizada", "pop_2022"]
dados.columns = colunas

for c in ["cod_uf", "pop_2010_sinopse", "pop_2010_compatibilizada", "pop_2022"]:
    dados[c] = pd.to_numeric(dados[c], errors="coerce")

dados.head(10)
```

```python
# 5 - Tabela de população agregada por estado
pop_por_estado = (
    dados.groupby("uf")[["pop_2010_compatibilizada", "pop_2022"]]
    .sum()
    .reset_index()
)
pop_por_estado
```

```python
# 6 - Organizar por população que mais cresceu entre 2010 e 2022 (estado)
pop_por_estado["crescimento_absoluto"] = pop_por_estado["pop_2022"] - pop_por_estado["pop_2010_compatibilizada"]
pop_por_estado = pop_por_estado.sort_values("crescimento_absoluto", ascending=False).reset_index(drop=True)
pop_por_estado
```

```python
# 7 - Salvar tabela por estado em csv
nome_arquivo_estado = "populacao_por_estado_crescimento"
caminho_estado = os.path.join(os.getcwd(), f"{nome_arquivo_estado}.csv")
pop_por_estado.to_csv(caminho_estado, sep=";", index=False)
```

```python
# 8 - Trabalhar por município
pop_por_municipio = dados[["uf", "municipio", "pop_2010_compatibilizada", "pop_2022"]].copy()
pop_por_municipio.head(10)
```

```python
# 9 - Organizar por população que mais cresceu entre 2010 e 2022 (município)
pop_por_municipio["crescimento_absoluto"] = pop_por_municipio["pop_2022"] - pop_por_municipio["pop_2010_compatibilizada"]
pop_por_municipio = pop_por_municipio.sort_values("crescimento_absoluto", ascending=False).reset_index(drop=True)
pop_por_municipio.head(10)
```

```python
# 10 - Salvar tabela por município em csv
nome_arquivo_municipio = "populacao_por_municipio_crescimento"
caminho_municipio = os.path.join(os.getcwd(), f"{nome_arquivo_municipio}.csv")
pop_por_municipio.to_csv(caminho_municipio, sep=";", index=False)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, ax = plt.subplots(figsize=(9, 8))
ordem = pop_por_estado.sort_values("pop_2022", ascending=False)
sns.barplot(data=ordem, y="uf", x="pop_2022", ax=ax, color="steelblue")
ax.set_title("População 2022 por UF")
ax.set_xlabel("Habitantes")
ax.set_ylabel("UF")
plt.tight_layout()
plt.show()
```
