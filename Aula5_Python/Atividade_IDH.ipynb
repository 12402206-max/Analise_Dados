```python
import pandas as pd

import matplotlib.pyplot as plt
```

```python
path = "Tabela4.csv"

bruto = pd.read_csv(path, sep=";", decimal=",", skiprows=1, encoding="latin1")

bruto
```

```python
indicador_5 = bruto.dropna(axis=1, how="all")

indicador_5
```

```python
idh_2024 = indicador_5[["Sigla", "Estado", "2024"]].sort_values("2024", ascending=False).reset_index(drop=True)

idh_2024
```

```python
indicador_5["melhora_idh"] = indicador_5["2024"] - indicador_5["1991"]

maior_melhora = indicador_5.loc[indicador_5["melhora_idh"].idxmax(), ["Sigla", "Estado", "1991", "2024", "melhora_idh"]]

print(f"Estado com a maior melhora: {maior_melhora['Estado']} ({maior_melhora['Sigla']})")
print(f"IDH em 1991: {maior_melhora['1991']:.3f}")
print(f"IDH em 2024: {maior_melhora['2024']:.3f}")
print(f"Melhora: {maior_melhora['melhora_idh']:.3f}")
```

```python
pioraram = indicador_5.loc[indicador_5["melhora_idh"] < 0, ["Sigla", "Estado", "1991", "2024", "melhora_idh"]]

if pioraram.empty:
    print("Nenhum estado teve piora de IDH entre 1991 e 2024.")
else:
    pioraram
```

```python
anos = [c for c in indicador_5.columns if str(c).isdigit()]

print(anos)

id_vars = [c for c in indicador_5.columns if c not in anos and c != "melhora_idh"]

df_longo = indicador_5.melt(
    id_vars=id_vars,
    value_vars=anos,
    var_name="Ano",
    value_name="IDH",
)

df_longo["Ano"] = df_longo["Ano"].astype(int)

df_longo["IDH"] = pd.to_numeric(df_longo["IDH"], errors="coerce")

df_longo.head()
```

```python
minas = df_longo[df_longo["Sigla"] == "MG"].sort_values("Ano")

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(minas["Ano"], minas["IDH"], marker="o", color="steelblue")

ax.set_title("Evolução do IDH em Minas Gerais (1991–2024)")

ax.set_xlabel("Ano")

ax.set_ylabel("IDH")

ax.set_ylim(0.3, 0.9)

plt.tight_layout()

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12, 7))

for sigla, grupo in df_longo.groupby("Sigla"):
    grupo = grupo.sort_values("Ano")
    ax.plot(grupo["Ano"], grupo["IDH"], marker="o", markersize=3, linewidth=1.5, label=sigla)

ax.set_title("Evolução do IDH por estado (1991–2024)")

ax.set_xlabel("Ano")

ax.set_ylabel("IDH")

ax.set_ylim(0.3, 0.9)

ax.legend(ncol=3, bbox_to_anchor=(1.02, 1), loc="upper left", title="UF")

fig.tight_layout()

plt.show()
```
